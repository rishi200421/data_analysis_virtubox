```python
import pandas as pd

bins = [0, 18, 25, 35, 45, 60, 100]
labels = ['Under 18', '18-24', '25-34', '35-44', '45-59', '60+']

df['Age_Group'] = pd.cut(
    df['Customer_Age'],
    bins=bins,
    labels=labels,
    right=False
)

# Count customers in each age group
df['Age_Group'].value_counts().sort_index()
```
